# Missing Value Treatment

```python
# since these features have a lot of null values, I choose to drop them
manipal_rains.drop(columns=['Evaporation', 'Sunshine', 'Cloud9am', 'Cloud3pm'], inplace=True)
```

```python
# from sklearn.impute import KNNImputer
#
# # I try KNN-imputing the rest
# imputer = KNNImputer()
# imputer.fit_transform(manipal_rains.select_dtypes('number'))
```

```python
# KNN-imputing takes an unrealistically long amount of time, so I use median
for i in manipal_rains.select_dtypes('number').columns:
    manipal_rains[i] = manipal_rains[i].fillna(manipal_rains[i].median())
```

```python
# Since RainToday and RainTomorrow are binary, I choose to fill nulls with the mode of the feature, which is 'No'
manipal_rains.RainToday = manipal_rains.RainToday.fillna(manipal_rains.RainToday.mode()[0])
manipal_rains.RainTomorrow = manipal_rains.RainTomorrow.fillna(manipal_rains.RainTomorrow.mode()[0])

manipal_rains[['WindGustDir', 'WindDir9am', 'WindDir3pm']]\
    = manipal_rains[['WindGustDir', 'WindDir9am', 'WindDir3pm']].bfill().ffill()
```

# Outlier Treatment & Misc

```python
# Interquartile-based clipping of numerical columns
for i in manipal_rains.select_dtypes('number').columns:
    q1, q3 = manipal_rains[i].quantile([.25, .75])
    iqr = q3 - q1
    manipal_rains[i] = manipal_rains[i].clip(q1 - iqr*1.5, q3 + iqr*1.5)
```

```python
# since Date has no null values, I attempt to parse it
manipal_rains.Date = pd.to_datetime(manipal_rains.Date, format='%d-%m-%Y')
```

# Numeric Encoding

```python
# Label encoding (binary)
manipal_rains.RainToday = manipal_rains.RainToday.astype('category').cat.codes
manipal_rains.RainTomorrow = manipal_rains.RainToday.astype('category').cat.codes
```

```python
# One Hot encoding
manipal_rains = pd.get_dummies(manipal_rains,
                       columns=['Location', 'WindGustDir', 'WindDir9am', 'WindDir3pm'],
                       dtype=int, drop_first=True)
```

```python
# Date encoding
manipal_rains['Year'] = manipal_rains.Date.dt.year
manipal_rains['Month'] = manipal_rains.Date.dt.month
manipal_rains['Day'] = manipal_rains.Date.dt.day
manipal_rains.drop(columns=['Date'], inplace=True)
```

# Scaling & Power Transformation

> Application of any of these depends on the algorithm being used

```python
from mlxtend.preprocessing import minmax_scaling

# Normalization

# print(*[i for i in list(manipal_rains.columns) if not i.startswith('Location')])
scale_columns = ['MinTemp', 'MaxTemp', 'Rainfall', 'WindGustSpeed', 'WindSpeed9am', 'WindSpeed3pm', 'Humidity9am', 'Humidity3pm', 'Pressure9am', 'Pressure3pm', 'Temp9am', 'Temp3pm']
manipal_rains[scale_columns] = minmax_scaling(manipal_rains[scale_columns], columns=scale_columns)
```

```python
# Standardization

scale_columns = ['MinTemp', 'MaxTemp', 'Rainfall', 'WindGustSpeed', 'WindSpeed9am', 'WindSpeed3pm', 'Humidity9am', 'Humidity3pm', 'Pressure9am', 'Pressure3pm', 'Temp9am', 'Temp3pm']
for i in scale_columns:
    manipal_rains[i] = (manipal_rains[i] - manipal_rains[i].mean()) / manipal_rains[i].std()
```

```python
from scipy.stats import boxcox

# Power Transformation

norm_columns = ['Rainfall', 'WindGustSpeed', 'WindSpeed9am', 'WindSpeed3pm']
# Idk if there's any need to normalize the humidity features

# some columns may have 0 values, so I'm adding a small constant
for i in norm_columns:
    if np.any(manipal_rains[i] == 0):
        manipal_rains[i] += 1

for i in norm_columns:
    manipal_rains[i] = boxcox(manipal_rains[i])[0]
```
